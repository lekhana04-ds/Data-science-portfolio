```python
import tensorflow as tf
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
shape = tensor.shape
print('tensor:', tensor)
print('Tensor shape:', shape)
```

```python
import tensorflow as tf
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])
size = tf.size(tensor)
print("Tensor size:", size)
```

```python
c = tf.constant([1, 2, 3])
print("Constant Tensor:", c.numpy())

v = tf.Variable([1, 2, 3])
print("Variable Tensor:", v.numpy())

z = tf.zeros([2, 3])
print("Zeros Tensor:/n", z.numpy())

o = tf.ones([2, 2])
print("Ones Tensor:/n", o.numpy())

rn = tf.random.normal([2, 3], mean=0.0, stddev=1.0)
print("Random Normal Tensor:\n", rn.numpy())

ru = tf.random.uniform([2, 2], minval=0, maxval=1.0)
print("random Uniform Tensor:\n", ru.numpy())
```

```python
import tensorflow as tf
print("Add:", tf.add(5, 3).numpy())
print("Subtract:", tf.subtract(10, 4).numpy())
print("Multiply:", tf.multiply([1, 2], [3, 4]).numpy())
print("Divide:", tf.divide([10, 20], [2, 5]).numpy())

a = tf.constant([[1, 2]])
b = tf.constant([[3],
                 [4]])
print("Matrix Multiplication:", tf.matmul(a, b).numpy())
print("Reduce Sum:", tf.reduce_sum([1, 2, 3]).numpy())
print("Reduce Mean:", tf.reduce_mean([2.0, 4.0, 6.0]).numpy())
```
