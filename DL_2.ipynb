```python
import numpy as np
input = np.array([5, 8])
weights = np.array([0.7, 0.4])
bias = 2
def sigmoid(z):
  return 1 / (1+np.exp(-z))
weighted_sum = np.dot(input, weights) + bias
output = sigmoid(weighted_sum)
print("f:Weighted_sum(z):{weighted_sum}")
print(f"output after sigmoid:{output}")
```
